```python
#Imports
import pandas as pd
import os
import period_reporting as pr #Be sure this .py file is in your local working directory (or import accordingly)
```

```python
#Review xlsx workflow files from working directory 
loc = './data' #<<< Change to 'mock_data' if testing this all out from you first pull (see README file)
xlsx_files = [loc+'/'+f for f in os.listdir(loc) if f.endswith('.xlsx')]
xlsx_files
```

```python
#select data source file
xlsx_file = pd.ExcelFile(xlsx_files[0]) # <<< Select the file you want to read in...

#Review sheet names (if multiple)
xlsx_file.sheet_names
```

```python
#select data source sheet
sheet = xlsx_file.sheet_names[6]
sheet 
```

```python
#create workflow df
df = pd.read_excel(
    xlsx_file, 
    sheet_name=sheet, 
    skiprows=1            # Skips the first row
)

df
```

```python
report_title = "All Issues - 2026"
select_columns = ['Key','Title/ Short Description','Date Started','Completion Date']
workflow_df = pr.period_workflows_reporter(
    df = df,                  # [OUTDATED] >> xlsx_file_location='.',
    period_code='Q',             # pandas frequency alias (e.g. 'D','W','W-MON','M','Q','Y',etc.)
    age_code='D',
    selected_standard_columns=select_columns,
    report_title= report_title,
    top_n_periods=7,
    verbose=True,
    #week_anchor='MON',            # optional day name to anchor weeks when using 'W'
    include_current_period=False
)
workflow_df
```
